```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import scanpy as sc
import anndata
```

```python
sc.set_figure_params(figsize=(5,5), frameon=False)
sc.settings.verbosity = 3
```

```python
projdir = '/u/project/cluo/terencew/igvf/2023_YR2/snmCT/mc'
donors = list(np.loadtxt(f'{projdir}/txt/donors.txt', dtype=str))
tmp_order = ['Start', 'Sendai', 'Fail1', 'Delayed', 'Inter1', 'Inter2', 'Fail2', 'IPS']
clusts = [f'{x}_{y}' for x,y in zip(np.repeat(tmp_order, len(donors)), np.tile(donors, len(tmp_order)))]
```

```python
adata = sc.read_h5ad(f'{projdir}/h5ad/integrate_m3c/merged.raw.h5ad')
```

```python
adata
```

### do the embedding per donor

```python
s = donors[0]
donor_adata = adata[adata.obs['line'] == s]
```

```python
hvfs = 20000
sc.pp.highly_variable_genes(adata, n_top_genes = hvfs)
```

```python
sc.tl.pca(adata, n_comps = 50, use_highly_variable=True)
```

```python
sc.pl.pca(adata, components = ['1,2', '3,4', '5,6', '7,8'], color=['time', 'line', 'assay'])
```

```python
sc.pp.neighbors(adata, n_pcs=20, random_state=0)
sc.tl.leiden(adata, resolution=0.3, random_state=0)
sc.tl.umap(adata, random_state=0)
```

```python
sc.pl.umap(adata, color=['time', 'line', 'assay'], ncols=2)
```

```python
adata.write_h5ad(f'{projdir}/h5ad/integrate_m3c/merged.processed.h5ad')
```
